```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

print("="*50)
print("TỔNG HỢP VISUALIZATION: MASTER DASHBOARD 2x2")
print("="*50)

image_paths = [
    '../outputs/figures/gradient_descent_convergence.png', 
    '../outputs/figures/Age_stroke_dist.png',
    '../outputs/figures/Correlation_heatmap.png',
    '../outputs/figures/Glucose_boxplot.png'
]

fig, axes = plt.subplots(2, 2, figsize=(20, 18))
fig.suptitle("STROKE ANALYSIS SUMMARY DASHBOARD", fontsize=24, fontweight='bold', y=0.98, color='#1a365d')

titles = [
    "1. TOÁN HỌC: Hội tụ của Gradient Descent", 
    "2. THỐNG KÊ MÔ TẢ: Phân phối Tuổi và Đột quỵ", 
    "3. CHUYÊN SÂU: Ma trận tương quan",
    "4. CHUYÊN SÂU: Phân bố Đường huyết"
]

axes_flat = axes.flatten()

for ax, img_path, title in zip(axes_flat, image_paths, titles):
    if os.path.exists(img_path):
        img = mpimg.imread(img_path)
        ax.imshow(img)
        ax.axis('off')
        ax.set_title(title, fontsize=16, fontweight='bold', pad=15)
    else:
        ax.text(0.5, 0.5, f"Không tìm thấy ảnh:\n{img_path}\nHãy kiểm tra lại tên file!", 
                ha='center', va='center', fontsize=12, color='red')
        ax.axis('off')

plt.tight_layout(rect=[0, 0, 1, 0.95])
os.makedirs('../outputs/figures', exist_ok=True)
plt.savefig('../outputs/figures/Stroke_Analysis_Summary.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n=> Đã xuất thành công file ảnh tổng hợp tại: ../outputs/figures/Stroke_Analysis_Summary.png")
```
